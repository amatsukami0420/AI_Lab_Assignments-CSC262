# Question 1: Constraint Satisfaction Problem (CSP)
Variables:
## $C = \{\text{AI}, \text{DS}, \text{CN}\}
$Domains:
## $D_{\text{AI}} = \{\text{Room B}, \text{Room C}\}$ (Room A excluded by Rule 1)$D_{\text{DS}} = \{\text{Room A}, \text{Room B}\}$ (Room C excluded by Rule 2)$D_{\text{CN}} = \{\text{Room A}, \text{Room C}\}$ (Restricted by Rule 4)Constraints:$\text{AI} \neq \text{Room A}$$\text{DS} \neq \text{Room C}$$\text{AllDiff}(C)$ (No two variables can have the same value)$\text{CN} \in \{\text{Room A}, \text{Room C}\}$

```python
# ==========================================
# Question 1: CSP using Backtracking Search
# ==========================================

def is_consistent(course, room, assignment):
    """
    Checks if assigning a room to a course violates any constraints.
    """
    # Constraint 1: AI cannot be assigned to Room A.
    if course == 'AI' and room == 'Room A':
        return False

    # Constraint 2: DS cannot be assigned to Room C.
    if course == 'DS' and room == 'Room C':
        return False

    # Constraint 4: CN must be assigned to either Room A or Room C.
    if course == 'CN' and room not in ['Room A', 'Room C']:
        return False

    # Constraint 3: No two courses can be assigned to the same classroom.
    if room in assignment.values():
        return False

    return True

def backtrack_search(assignment, variables, domains):
    """
    Recursive Backtracking Search Algorithm to find a valid assignment.
    """
    # Base Case: If all variables are assigned, return the assignment
    if len(assignment) == len(variables):
        return assignment

    # Select an unassigned variable
    unassigned_vars = [v for v in variables if v not in assignment]
    current_var = unassigned_vars[0]

    # Iterate through the domain (possible rooms) of the current variable
    for room in domains[current_var]:
        # Check if the room assignment is valid under current constraints
        if is_consistent(current_var, room, assignment):
            assignment[current_var] = room

            # Recursively assign the rest
            result = backtrack_search(assignment, variables, domains)
            if result is not None:
                return result

            # Backtrack if no valid assignment is found down this path
            del assignment[current_var]

    return None

# Problem Setup
courses = ['AI', 'DS', 'CN']

# Initializing domains with all possible rooms for simplicity;
# the is_consistent function handles the logic constraints.
available_rooms = {
    'AI': ['Room A', 'Room B', 'Room C'],
    'DS': ['Room A', 'Room B', 'Room C'],
    'CN': ['Room A', 'Room B', 'Room C']
}

# Execute Search
solution = backtrack_search({}, courses, available_rooms)

print("--- Valid Classroom Assignment ---")
if solution:
    for course, room in solution.items():
        print(f"Course: {course} -> {room}")
else:
    print("No valid assignment found.")
```

# Question 2: Knowledge-Based System (KBS).

```python
# ==========================================
# Question 2: Knowledge-Based System
# ==========================================

class GradingKBS:
    def __init__(self):
        # Define the knowledge base as a list of dictionary rules
        self.knowledge_base = [
            {"rule": "Marks > 85", "condition": lambda m: m > 85, "conclusion": "A"},
            {"rule": "Marks > 70 and Marks <= 85", "condition": lambda m: 70 < m <= 85, "conclusion": "B"},
            {"rule": "Marks > 50 and Marks <= 70", "condition": lambda m: 50 < m <= 70, "conclusion": "C"},
            {"rule": "Marks <= 50", "condition": lambda m: m <= 50, "conclusion": "Fail"}
        ]

    def evaluate_student(self, marks):
        """
        Applies the rules in the knowledge base to determine the grade.
        """
        # Inference engine: checking facts against rules
        for rule in self.knowledge_base:
            if rule["condition"](marks):
                return rule["conclusion"]
        return "Invalid Input"

# Initialize the system
kbs = GradingKBS()

# Task 3: Test program using at least three different inputs
test_cases = [92, 78, 65, 42]

print("--- KBS Student Grade Evaluation ---")
for mark in test_cases:
    grade = kbs.evaluate_student(mark)
    print(f"Student Marks: {mark} | Academic Grade: {grade}")

# Optional: Uncomment below to allow manual input testing in Colab
# try:
#     user_marks = float(input("\nEnter student marks to test: "))
#     print(f"Assigned Grade: {kbs.evaluate_student(user_marks)}")
# except ValueError:
#     print("Please enter a valid numeric value.")
```
